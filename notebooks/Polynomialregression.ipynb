Polynomial regression 

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import joblib
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('polynomial.csv')
```

```python
df
```

```python
X=df[['Speed']]
Y=df[['MPG']]
```

```python
poly=PolynomialFeatures(degree=2)
xoply=poly.fit_transform(X)
```

```python
model=LinearRegression()
```

```python
model.fit(xoply,Y)
```

```python
plt.figure(figsize=(10,10))
plt.scatter(X,Y)
plt.plot(X,model.predict(xoply))
plt.plot()
```

```python
model.intercept_
```

```python
joblib.dump(model,'../backend/fuel_model.pkl')
joblib.dump(poly,'../backend/poly_transformer.pkl')
print(f"Intercept: {model.intercept_} and Cofficient: {model.coef_[0]}")
```
